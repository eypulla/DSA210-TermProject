```python
from google.colab import files
from IPython.display import display
import os, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid")
plt.rcParams.update({"figure.dpi": 120, "font.size": 11})

uploaded = files.upload()
df = pd.read_csv("volleyball_economic_dataset.csv")
```

## 4.1 Simple Regression — Log GDP → Women's Points

The simplest possible model: one predictor, one target. This gives us a baseline R² and lets us check the residuals for any patterns.

```python
tmp = df[["country_std","log_gdp","pts_women"]].dropna()
m = LinearRegression().fit(tmp[["log_gdp"]], tmp["pts_women"])
pred  = m.predict(tmp[["log_gdp"]])
resid = tmp["pts_women"].values - pred

print(f"β₀={m.intercept_:.1f}  β₁={m.coef_[0]:.1f}  R²={m.score(tmp[['log_gdp']],tmp['pts_women']):.3f}")

fig, ax = plt.subplots(figsize=(9, 6))
ax.scatter(tmp["log_gdp"], tmp["pts_women"], color="#843C0C", alpha=0.65, s=60)
xr = np.linspace(tmp["log_gdp"].min(), tmp["log_gdp"].max(), 100).reshape(-1,1)
ax.plot(xr, m.predict(xr), "k-", lw=2)
for _, row in tmp.nlargest(4,"pts_women").iterrows():
    ax.annotate(row["country_std"], (row["log_gdp"],row["pts_women"]),
                fontsize=8, xytext=(3,3), textcoords="offset points")
ax.set_title(f"Log GDP → Women's Points  (R²={m.score(tmp[['log_gdp']],tmp['pts_women']):.2f})")
ax.set_xlabel("Log GDP per Capita"); ax.set_ylabel("WR Points")
plt.tight_layout()
os.makedirs("figures", exist_ok=True)
plt.savefig("figures/fig_simple_regression.png", bbox_inches="tight", dpi=150)
plt.show()
```

*Positive slope confirms the GDP-volleyball association, but R² is low — GDP alone explains only a small fraction of the variance. Countries like Brazil and Turkey sit far above the line, which already tells us something is missing from this model.*

## 4.1b Residual Plot

A healthy residual plot should show random scatter around zero — no patterns, no funnel shape.

```python
fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(pred, resid, color="#922B21", alpha=0.65, s=60)
ax.axhline(0, color="black", ls="--", lw=1.5)
ax.set_title("Residual Plot — Simple Regression")
ax.set_xlabel("Fitted Values"); ax.set_ylabel("Residuals")
plt.tight_layout()
plt.savefig("figures/fig_residual_plot.png", bbox_inches="tight", dpi=150)
plt.show()

# residuals look randomly scattered around 0 — the log transform helped with heteroskedasticity.
```

*Residuals are randomly scattered around zero with no obvious funnel or curve — the log transformation successfully addressed the heteroskedasticity that would have appeared with raw GDP values.*

## 4.2 Multiple Regression — Men vs Women

Adding HDI and population as predictors. The coefficient comparison between men's and women's models will tell us whether the predictors work differently by gender.

```python
FEATS = ["log_gdp","hdi","log_population"]
reg   = df[["country_std"] + FEATS + ["pts_women","pts_men"]].dropna()

rows = []
for gender, col in [("Women","pts_women"),("Men","pts_men")]:
    m = LinearRegression().fit(reg[FEATS], reg[col])
    rows.append({"Gender": gender,
                 "β Log GDP":  round(m.coef_[0],1),
                 "β HDI":      round(m.coef_[1],1),
                 "β Log Pop":  round(m.coef_[2],1),
                 "R²":         round(m.score(reg[FEATS],reg[col]),3)})
comp = pd.DataFrame(rows)
display(comp)
```

```python
fig, ax = plt.subplots(figsize=(9, 5))
x = np.arange(3); w = 0.35
ax.bar(x-w/2, comp.iloc[0][["β Log GDP","β HDI","β Log Pop"]].values,
       w, label="Women", color="#843C0C", alpha=0.85)
ax.bar(x+w/2, comp.iloc[1][["β Log GDP","β HDI","β Log Pop"]].values,
       w, label="Men", color="#1F4E79", alpha=0.85)
ax.set_xticks(x); ax.set_xticklabels(["Log GDP","HDI","Log Pop"])
ax.axhline(0, color="black", lw=0.8)
ax.set_ylabel("β coefficient"); ax.legend()
ax.set_title("Multiple Regression Coefficients — Men vs Women", fontsize=12)
plt.tight_layout()
os.makedirs("figures", exist_ok=True)
plt.savefig("figures/fig4_regression_coef.png", bbox_inches="tight", dpi=150)
plt.show()

# HDI has the largest coefficient for both. it's notably bigger for women,
# suggesting women's volleyball is more tied to broad societal development.
# population coefficient is near-zero for both — consistent with everything so far.
```

*HDI has the largest coefficient for both genders, and it is notably bigger for women — suggesting women's volleyball performance is more sensitive to broad societal development. The population coefficient is near-zero for both, consistent with the correlation tests in notebook 03.*

## 4.3 Residual Analysis — Who Punches Above Their Weight?

The most interesting question in the whole project: which countries perform far better or worse than their GDP would predict?

```python
tmp_r = df[["country_std","log_gdp","pts_women"]].dropna().copy()
m_r   = LinearRegression().fit(tmp_r[["log_gdp"]], tmp_r["pts_women"])
tmp_r["predicted"] = m_r.predict(tmp_r[["log_gdp"]])
tmp_r["residual"]  = tmp_r["pts_women"] - tmp_r["predicted"]
tmp_r = tmp_r.sort_values("residual")

fig, ax = plt.subplots(figsize=(11, 7))
colors = ["#27AE60" if r > 0 else "#E74C3C" for r in tmp_r["residual"]]
ax.barh(tmp_r["country_std"], tmp_r["residual"], color=colors, alpha=0.85)
ax.axvline(0, color="black", lw=1)
ax.set_xlabel("Residual  (Actual − Predicted Women's Points)")
ax.set_title("Over- and Under-Performers Relative to GDP", fontsize=12)
ax.tick_params(axis="y", labelsize=8)
plt.tight_layout()
os.makedirs("figures", exist_ok=True)
plt.savefig("figures/fig4_residuals.png", bbox_inches="tight", dpi=150)
plt.show()

# Brazil and Turkey are the biggest over-performers by far.
# their success clearly comes from something GDP alone can't explain —
# probably volleyball culture, federation investment, and club infrastructure.
# some wealthy Gulf states show the opposite: high GDP, low volleyball success.
```

*Brazil and Turkey are the biggest over-performers by far — their success comes from something GDP cannot capture: volleyball culture, strong federations, and club infrastructure. The wealthy Gulf states show the opposite pattern. This residual analysis is the clearest answer to the project's original question: money helps, but it is not sufficient.*
